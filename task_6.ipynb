```python
#------Task 6:Number Gussing Game---
import random

secret=random.randint(1,20) #secret number in [1,20]
print("I'm thinking of a number between 1 to 20......")

while True:
    guess=int(input("Your guessed:"))

    if guess==secret:
        print("Correct! you guessed it.")
        break
    elif guess<secret:
        print("Too low,try again.")
    else:
        print("Too High,try again.")
```
